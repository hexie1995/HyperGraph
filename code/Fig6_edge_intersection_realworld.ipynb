```python
def plot_intersection(data, timesteps, name):
    
    
    for i in range(0, len(toplot[10])):
        k0 = [x[i] for x in data[10:]]
        plt.plot(range(len(data[10:])), k0, label = "k = " + str(i))
        
    plt.legend()
    plt.loglog()
    plt.title(name)
    plt.xlabel("Timesteps")
    plt.ylabel("Probablity of $r_{k}$")
    plt.savefig("figures/realworld_intersection/{}.pdf".format(name))
    plt.show()
    plt.close()
```

```python
#for data in ["email-enron"]:
for data in realworld_Hgraphs:
    timesteps = timestamp_dict[data]
    try:
        for g in Gtypes:
            toplot = np.load('real_simulation/' + data + '_' + g + ".npy", allow_pickle=True) 
            plot_intersection(toplot, timesteps,  data + '_' + g)
    except:
        print(data, "not finished")
```
