# Computing the Intersection Array

This notebook demonstrates the use of `GrowingHypergraph.intersection_array()` to compute an array counting intersections between edges. Doing this exactly is computationally very expensive and so the supplied implementation works by sampling: 

- A 3-dimensional array `P` is initialized. 
- For a user-specified number of iterations: 
    - A pair of edges is sampled uniformly at random. 
    - The sizes `i` and `j` of those edges are computed. 
    - The size `k` of the *intersection* between those edges are computed. 
    - `P[i,j,k] += 1` and `P[j,i,k] += 1`. 

This array can be normalized in various ways. For example, supplying the argument `normalize = True` will make it so that `P[i,j,k]` corresponds to $q_{k|ij}$, the probability that two edges of given sizes $i$ and $j$ have an intersection of size $k$:  

```python
%load_ext autoreload
%autoreload 2
import importlib 
m = importlib.import_module(".model", "src")
from matplotlib import pyplot as plt
import numpy as np
import xgi
import copy 
plt.style.use("seaborn-whitegrid")

import matplotlib as mpl
from matplotlib.colors import LinearSegmentedColormap, ListedColormap
import cProfile
```

```python
dataset = "email-enron"
H = xgi.load_xgi_data(dataset)
GH = m.GrowingHypergraph(H)
```

Let's compute an intersection array. On the head node of Middlebury's HPC cluster, this took about 30 seconds. 

```python
P = GH.intersection_array(num_samples = int(1e7), normalize = False)
```

Plot of the different slices: we look at all the ways to form intersections of size $k$, for varying $k$: 

```python
fig, axarr = plt.subplots(2, 5, figsize = (10, 4), sharex = True, sharey = True)

for i in range(10):
    ax = axarr.ravel()[i]
    ax.imshow(P[:11, :11, i+1], zorder = 10)
    # ax.axis("off")
    if i in [0, 5]:
        ax.set(ylabel = r"$i$")
    
    if i >= 5: 
        ax.set(xlabel = r"$j$")
        
    ax.set(title = f"{i+1}")
    plt.tight_layout()
    fig.suptitle("Intersections of size...")
```

We could instead compute the normalized version: 

```python
P = GH.intersection_array(num_samples = int(1e7), normalize = True)
```

```python
i = 3
j = 4
k = 2

q = P[i, j, k]
print(f"The probability that a random edge of size {i} and a random edge of \nsize {j} have an intersection of size {k} is approximately {q:.2f}.")
```
