```python
import xgi
import numpy as np

def model_name_tbd(H, z, p, q, s = 0):
    
# Parameter 1 z = Number of Newly appeared Node in the new temporal layer
# Parameter 2 p = probability node removal from original set
# Parameter 3 q = probability of node addition to the original set
# parameter 4 s = total changed number of edges in t+1, could be positive or negative.
 
    H_new = xgi.Hypergraph()
    edge_dict = xgi.convert.to_hyperedge_dict(H)
    edgelist =list(edge_dict.values())
    new_edgelist = np.random.choice(edgelist, len(edgelist) + s)
    
    nodes = list(H.nodes)
    new_nodes = range(len(nodes) + z)
    node_key = nodes+ list(range(len(nodes),len(nodes) + z))
    node_set = set(node_key)
    
    
    node_dict = dict(zip(new_nodes, node_key))
    # add and remove nodes according to the parameters
    
    for edge_id in new_edgelist:
        
        # remove nodes from the current set
        curr_nodes = list(edge_id)
        to_be_removed = np.random.binomial(1,[p]*len(curr_nodes))
        removed_nodes = np.where(to_be_removed==1)
        
        edge_id.difference_update(set(np.array(curr_nodes)[removed_nodes]))
        # add new_nodes to the current set
        rest_nodes = list(node_set - edge_id)
        to_be_added = np.random.binomial(1,[q]*len(rest_nodes))
        added_nodes = np.where(to_be_removed==1)
        edge_id.difference_update(set(np.array(rest_nodes)[added_nodes]))
        #new_edge.remove(node_id)

        
    H_new.add_nodes_from(new_nodes)
    H_new.add_edges_from(new_edgelist)
    
    return H_new, node_dict
```
