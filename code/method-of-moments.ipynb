## Method of Moments

So, the big picture here is to try to use the method of moments to perform estimation for large hypergraphs. Preliminary experiments make it look like this works *sometimes* and that some systematic experiments should be done. 

The method of moments just involves numerically solving a system of 3 nonlinear equations, which can be quite fast. However, the `fsolve` call needs to be wrapped in a grid search in order to give valid solutions. Solutions are valid of all entries are nonnegative and $\eta \leq 1$.

```python
import importlib 
m = importlib.import_module(".model", "src")
em = importlib.import_module(".em", "src")
moments = importlib.import_module(".moments", "src")
from matplotlib import pyplot as plt
import numpy as np
from scipy.optimize import fsolve
from itertools import product
import pandas as pd 
```

```python
# generate some sample data

eta   = 0.7   # node retention in edges
gamma = 0.3   # poisson number of nodes from graph
beta  = 0.7   # poisson number of novel nodes

timesteps = int(5e4)

H = m.GrowingHypergraph()
H.add_edge((0, 1))
H.add_edge((2, 3))

for _ in range(timesteps):
    H.sample_edge(eta, gamma, beta, True)
```

```python
theta, obj = moments.method_of_moments_estimator(H)
```
