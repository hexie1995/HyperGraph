```python
import xgi
import numpy as np
```

```python
def edge_neighborhood(H, eid):
    """
    compute the edge ego-network of the edge with specified eid in hypergraph H
    """
    
    # get the nodes of the input edge id
    e = H.edges.members(eid)
    neighbor_edges = []
    
    # loop over the nodes in e, get all edges containing them, and add to neighbor_edges
    for i in e: 
        for eid_ in H.nodes.memberships(i):
            if (eid_ != eid): 
                neighbor_edges.append(eid_)
    
    # don't include the same edge multiple times
    neighbor_edges = set(neighbor_edges)
    
    # get the edges as sets, not just as ids
    neighbor_edges = [H.edges.members(eid_) for eid_ in neighbor_edges]
    
    # construct a new hypergraph and add the retrieved edges to it
    H_ = xgi.Hypergraph()
    
    for edge in neighbor_edges: 
        H_.add_edge(edge)
    
    return H_
    
```

```python
# illustration on enron email
H = xgi.load_xgi_data("email-enron")

# random edge
eid = np.random.choice(H.edges)

# edge ego network of random edge
H_ = edge_neighborhood(H, eid)

# illustrate
print(f"Original hypergraph has {len(H.nodes)} nodes and {len(H.edges)} edges.")
print(f"Edge ego network of edge {eid} has {len(H_.nodes)} nodes and {len(H_.edges)} edges.")
```
