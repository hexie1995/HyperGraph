# Baby Degree Distribution Experiment

This is a minimal experiment (with a minimal, VERY SLOW model implementation) to inform study of the degree distribution of the proposed model. It's mostly just to help Phil to form conjectures about things to try to derive. It might also be useful for Xie in implementing a faster, more full-featured version of our model using XGI. As is, the model is pretty slow and that makes Phil sad.  

```python
from matplotlib import pyplot as plt
import numpy as np
import importlib 
m = importlib.import_module(".baby-model", "src")
```

```python
n = 10000
GAMMA = 5 # roughly, expected number of novel nodes per edge
eta = 0.5 # fraction of nodes retained in each new edge

gamma = GAMMA/n

H = m.GrowingHypergraph(N = range(n))
H.add_edge((0, 1))

for _ in range(50000):    
    H.sample_edge(eta, gamma)
```

```python
deg = H.degree_sequence()
```

```python
c, b = np.histogram(np.array(list(deg.values() )), bins = 30)
fig, ax = plt.subplots(1, 1)
ax.scatter(b[:-1], c)
ax.semilogy()
labs = ax.set(xlabel = "Node degree", ylabel = "Frequency", title = "Looks Poisson maybe?")
```
